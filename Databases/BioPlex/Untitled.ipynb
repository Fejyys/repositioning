```python
import csv

with open('BioPlex.tsv') as f_in, \
     open('sanitised_bioplex.csv', 'w', newline='') as f_out:

    reader = csv.reader(f_in, delimiter='\t', quotechar='"')
    writer = csv.writer(f_out)

    header = next(reader)
    idx_a = header.index("UniprotA")
    idx_b = header.index("UniprotB")

    writer.writerow(["Uniprot_A", "Uniprot_B"])

    for row in reader:
        writer.writerow([row[idx_a], row[idx_b]])
```
