```python
import csv

def extract_uniprot(alt_ids_field):
    for entry in alt_ids_field.split('|'):
        if entry.startswith('uniprot/swiss-prot:'):
            return entry.split(':', 1)[1]
    return None
def extract_confidence(conf_value):
    if conf_value.startswith('score:'):
        return conf_value.split(':',1)[1]
    return '-'
with open('E:/human/BIOGRID-ORGANISM-Homo_sapiens-5.0.260.mitab.txt') as f_in, \
     open('uniprot_edges.csv', 'w', newline='') as f_out:

    reader = csv.reader(f_in, delimiter='\t')
    writer = csv.writer(f_out)

    header = next(reader)
    writer.writerow(["Uniprot_A", "Uniprot_B", "Confidence Values"])

    for row in reader:
        alt_a, alt_b, conf = row[2], row[3], row[14]
        uniprot_a = extract_uniprot(alt_a)
        uniprot_b = extract_uniprot(alt_b)
        conf = extract_confidence(conf)
        if uniprot_a and uniprot_b:
            writer.writerow([uniprot_a, uniprot_b, conf])
            
```
