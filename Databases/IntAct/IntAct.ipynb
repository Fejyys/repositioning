```python
import csv
import pandas as pd
df = pd.read_csv("E:/human/human.txt", sep="\t", header=1, usecols=[0, 1])
interactors = df.iloc[:, [0, 1]]
interactors.columns = ["Uniprot_A", "Uniprot_B"]

interactors.to_csv("sanitised.tsv", sep="\t", index=False, header=True)
```

```python

df = pd.read_csv("sanitised.tsv", sep="\t", header=1)
df.columns = ["Uniprot_A", "Uniprot_B"]
df["Uniprot_A"] = df["Uniprot_A"].str.replace("uniprotkb:", "", regex=False)
df["Uniprot_B"] = df["Uniprot_B"].str.replace("uniprotkb:", "", regex=False)
print(df[["Uniprot_A", "Uniprot_B"]].head())
df.to_csv("sanitised.tsv", sep="\t", index=False, header=True)
```
